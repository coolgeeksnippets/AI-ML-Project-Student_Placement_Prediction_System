```python
y = df_clean['placement_status'].map({
    'Placed': 1,
    'Not Placed': 0
})

print(y.head())
```

```python
X = df_clean.drop(
    columns=['placement_status', 'placement_status_encoded']
)

print("Input features created successfully!")
print("Shape of X:", X.shape)
```

```python
print("Shape of X:", X.shape)
print("Shape of y:", y.shape)

print("\nFeatures in X:")
print(X.columns.tolist())

print("\nTarget name:", y.name)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training and testing data created successfully!")
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
print("Training set placement distribution:")
print(y_train.value_counts())

print("\nTraining set placement percentage:")
print(y_train.value_counts(normalize=True) * 100)
```

```python
print("Testing set placement distribution:")
print(y_test.value_counts())

print("\nTesting set placement percentage:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
print("Total training samples:", len(X_train))
print("Total testing samples:", len(X_test))
print("Total samples:", len(X_train) + len(X_test))
```

```python
numeric_features = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_features = X_train.select_dtypes(include=['object']).columns.tolist()

print("Numerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
numeric_transformer = StandardScaler()

categorical_transformer = OneHotEncoder(
    handle_unknown='ignore'
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

print("Training data preprocessing completed.")
print("Processed training shape:", X_train_processed.shape)
```

```python
X_test_processed = preprocessor.transform(X_test)

print("Testing data preprocessing completed.")
print("Processed testing shape:", X_test_processed.shape)
```

```python
print("Original number of features:", X_train.shape[1])
print("Processed training features:", X_train_processed.shape[1])
print("Processed testing features:", X_test_processed.shape[1])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
```

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train_processed, y_train)

print("Logistic Regression model trained successfully!")
```

```python
decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

decision_tree_model.fit(X_train_processed, y_train)

print("Decision Tree model trained successfully!")
```

```python
random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

random_forest_model.fit(X_train_processed, y_train)

print("Random Forest model trained successfully!")
```

```python
from sklearn.svm import LinearSVC

svm_model = LinearSVC(
    random_state=42,
    max_iter=5000
)

svm_model.fit(X_train_processed, y_train)

print("Support Vector Machine model trained successfully!")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
y_pred_logistic = logistic_model.predict(X_test_processed)
y_pred_tree = decision_tree_model.predict(X_test_processed)
y_pred_forest = random_forest_model.predict(X_test_processed)
y_pred_svm = svm_model.predict(X_test_processed)

print("Predictions generated successfully!")
```

```python
models = {
    'Logistic Regression': y_pred_logistic,
    'Decision Tree': y_pred_tree,
    'Random Forest': y_pred_forest,
    'SVM': y_pred_svm
}

results = []

for model_name, predictions in models.items():
    results.append({
        'Model': model_name,
        'Accuracy': accuracy_score(y_test, predictions),
        'Precision': precision_score(y_test, predictions),
        'Recall': recall_score(y_test, predictions),
        'F1 Score': f1_score(y_test, predictions)
    })

results_df = pd.DataFrame(results)

results_df
```

```python
results_df.set_index('Model')[['Accuracy', 'Precision', 'Recall', 'F1 Score']].plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Comparison of Machine Learning Models')
plt.xlabel('Model')
plt.ylabel('Score')
plt.xticks(rotation=0)
plt.ylim(0, 1)
plt.legend()
plt.show()
```

```python
best_model_name = results_df.loc[
    results_df['F1 Score'].idxmax(),
    'Model'
]

best_f1_score = results_df['F1 Score'].max()

print("Best model based on F1 Score:", best_model_name)
print("Best F1 Score:", best_f1_score)
```

```python
feature_importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': random_forest_model.feature_importances_
})
feature_importance_df = feature_importance_df.sort_values(
    by='Importance',
    ascending=False
)
feature_importance_df
```

```python
top_15_features.sort_values(
    by='Importance'
).plot(
    x='Feature',
    y='Importance',
    kind='barh',
    figsize=(10, 7),
    legend=False
)

plt.title('Top 15 Feature Importances for Placement Prediction')
plt.xlabel('Feature Importance')
plt.ylabel('Feature')
plt.show()
```

```python
original_feature_importance = {}

for feature, importance in zip(
    feature_names,
    random_forest_model.feature_importances_
):
    
    clean_feature = feature.replace('num__', '').replace('cat__', '')
    
    matched_feature = None
    
    for categorical_feature in categorical_features:
        if clean_feature.startswith(categorical_feature + '_'):
            matched_feature = categorical_feature
            break
    
    if matched_feature is None:
        matched_feature = clean_feature
    
    original_feature_importance[matched_feature] = (
        original_feature_importance.get(matched_feature, 0)
        + importance
    )

original_importance_df = pd.DataFrame(
    list(original_feature_importance.items()),
    columns=['Feature', 'Total Importance']
).sort_values(
    by='Total Importance',
    ascending=False
)

original_importance_df
```

```python
def get_risk_level(placement_probability):
    if placement_probability < 0.50:
        return "High Risk"
    elif placement_probability < 0.75:
        return "Medium Risk"
    else:
        return "Low Risk"
```

```python
risk_level = get_risk_level(placement_probability)
print("Prediction:", "Placed" if prediction == 1 else "Not Placed")
print(f"Placement Probability: {placement_probability:.2%}")
print("Placement Risk Level:", risk_level)
```
